```python
plt.figure(figsize=(15, 12))
sns.heatmap(dataset.corr(), annot=True, cmap='RdYlGn')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.countplot(x='Class', data=dataset)
plt.title('Distribution of Fraudulent vs Non-Fraudulent Transactions')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(dataset['Amount'], bins=50, kde=True)
plt.title('Distribution of Transaction Amount')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(np.log1p(dataset['Amount']), bins=50, kde=True)
plt.title('Log-Scaled Transaction Amount Distribution')
plt.show()
```

```python
sns.boxplot(x='Class', y='Amount', data=dataset)
plt.title('Transaction Amount per Class')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Time', y='Amount', hue='Class', data=dataset, alpha=0.6)
plt.title('Time vs Amount Colored by Class')
plt.show()
```

```python
corr_target = dataset.corr()['Class'].drop('Class')
corr_target.sort_values(ascending=False).plot(kind='bar', figsize=(10, 4))
plt.title('Correlation of Features with Class (Fraud)')
plt.show()
```

```python
# Sample 1000 rows for performance (adjust as needed)
sample_data = dataset.sample(n=1000, random_state=42)

# Only include numeric columns and the target
columns_to_plot = ['V1', 'V2', 'V3', 'V4', 'V5', 'Class']  # Choose any relevant subset
sns.pairplot(sample_data[columns_to_plot], hue='Class', diag_kind='kde', palette='husl')
plt.suptitle('Pairplot of Selected Features Colored by Class', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
for label in [0, 1]:
    subset = dataset[dataset['Class'] == label]
    sns.kdeplot(subset['V1'], label=f'Class {label}', fill=True)

plt.title('KDE Plot of Feature V1 by Class')
plt.xlabel('V1')
plt.ylabel('Density')
plt.legend()
plt.show()
```
